```python
import pandas as pd
from pathlib import Path
import json
from sklearn.model_selection import train_test_split
```

```python
ROOT = Path.cwd().parent
subject='DATA_RAW_MODELS/Subject_Data.csv'
motor='DATA_RAW_MODELS/Motor_Data.csv'
cognitive='DATA_RAW_MODELS/Cognitive_Data.csv'
sleep='DATA_RAW_MODELS/Sleep_Data.csv'
ledd='DATA_RAW_MODELS/LEDD_Medication_Data.csv'
```

# DATA SPLIT

```python
data_class=pd.read_csv(ROOT / subject)
data_class=data_class[['PATNO',"Visit ID",'UPDRS_III_ProgressionType']]
data_class.head(3)
```

```python
X=data_class[['PATNO','Visit ID']]
y=data_class['UPDRS_III_ProgressionType']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

Subjects_train = (X_train['PATNO'].astype(str) + '_' + X_train['Visit ID'].astype(str)).tolist()

Subjects_test = (X_test['PATNO'].astype(str) + '_' + X_test['Visit ID'].astype(str)).tolist()


data = {
    "Train_80": Subjects_train,
    "Test_20": Subjects_test
}

with open(ROOT / 'DATA_RAW_MODELS/Data_split.json', "w", encoding="utf-8") as archivo:
    json.dump(data, archivo, indent=4, ensure_ascii=False)
```

## Data Selection

```python
subject_data.drop(columns=['Sporadic PD at Enrollment',
                           'UPDRS_I_ProgressionType', 'UPDRS_II_ProgressionType',
                           'Identify self as Asian', 'Identify self as Black/African American', 
                           'Identify self as Hawaiian/Other Pacific Islander', 'Identify self as American Indian/Alaska Native',
                           'MDS-UPDRS Total Score 1YearProg', 'UPDRS_III_1YearProg',
                           'Δ TOTAL_UPDRS_1Y', 'Δ UPDRS_I_1Y', 'Δ UPDRS_II_1Y', 'Δ UPDRS_III_1Y', 'Δ UPDRS_IV_1Y','Delta_SBP',
                           'Delta_DBP', 'Delta_HR', 'Height (cm)', 'Weight (kg)',
                           'Temperature (Celsius)', 'Supine BP - systolic (mmHg)', 'Supine BP - diastolic (mmHg)', 
                           'Supine heart rate (bpm)', 'Standing BP - systolic (mmHg)', 'Standing BP - diastolic (mmHg)', 'Standing heart rate (bpm)',
                           'UPDRS_I_Class', 'UPDRS_II_Class', 'UPDRS_III_Class', 'UPDRS_IV_Class'
                           ], inplace=True)
subject_data.head(3)
```

```python
subject_data['UPDRS_III_ProgressionType'].value_counts()
```

```python
numeric_cols = subject_data.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = subject_data.select_dtypes(include=['object','bool']).columns.tolist()

#'Ortostatic_Hypotension'
subject_data['Ortostatic_Hypotension'] = subject_data['Ortostatic_Hypotension'].map({True: 1, False: 0})
# 'UPDRS_III_ProgressionType'
subject_data['UPDRS_III_ProgressionType'] = subject_data['UPDRS_III_ProgressionType'].map({'Substantial improv.': -1, 'No significant change.': 0,'Substantial progress.': 1})
subject_data.head(3)

print("Numeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)

```
