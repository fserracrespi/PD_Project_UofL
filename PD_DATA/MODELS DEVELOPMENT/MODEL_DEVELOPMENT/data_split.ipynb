```python
import pandas as pd
from pathlib import Path
import json
from sklearn.model_selection import train_test_split
```

```python
ROOT = Path.cwd().parent
subject='DATA_RAW_MODELS/Subject_Data.csv'
motor='DATA_RAW_MODELS/Motor_Data.csv'
cognitive='DATA_RAW_MODELS/Cognitive_Data.csv'
sleep='DATA_RAW_MODELS/Sleep_Data.csv'
ledd='DATA_RAW_MODELS/LEDD_Medication_Data.csv'
```

# DATA SPLIT

```python
data_class=pd.read_csv(ROOT / subject)
data_class=data_class[['PATNO',"Visit ID",'UPDRS_III_ProgressionType']]
data_class.head(3)
```

```python
X=data_class[['PATNO','Visit ID']]
y=data_class['UPDRS_III_ProgressionType']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

Subjects_train = (X_train['PATNO'].astype(str) + '_' + X_train['Visit ID'].astype(str)).tolist()

Subjects_test = (X_test['PATNO'].astype(str) + '_' + X_test['Visit ID'].astype(str)).tolist()


data = {
    "Train_80": Subjects_train,
    "Test_20": Subjects_test
}

with open(ROOT / 'DATA_RAW_MODELS/Data_split.json', "w", encoding="utf-8") as archivo:
    json.dump(data, archivo, indent=4, ensure_ascii=False)
```

# Subject Data

```python
subject_data['UPDRS_III_ProgressionType'].value_counts()
```

```python
#'Ortostatic_Hypotension'
subject_data['Ortostatic_Hypotension'] = subject_data['Ortostatic_Hypotension'].map({True: 1, False: 0})
# 'UPDRS_III_ProgressionType'
subject_data['UPDRS_III_ProgressionType'] = subject_data['UPDRS_III_ProgressionType'].map({'Substantial improv.': -1, 'No significant change.': 0,'Substantial progress.': 1})

subject_data['subject_visit'] = subject_data['PATNO'].astype(str) + '_' + subject_data['Visit ID'].astype(str)
subject_data.drop(columns=['PATNO','Visit ID'], inplace=True)

subject_data.head(3)
```

# Motor Data

```python
motor_data = pd.read_csv(ROOT / motor)
motor_data['subject_visit'] = motor_data['PATNO'].astype(str) + '_' + motor_data['Visit ID'].astype(str)
motor_data.drop(columns=['PATNO','Visit ID'], inplace=True)
motor_data.head(3)
```

## Data Download

```python
motor_data.to_csv(ROOT / 'DATA_MODEL_Dev/Motor_data_models.csv', index=False)
```

# Cognitive Data

```python
cognitive_data = pd.read_csv(ROOT / cognitive)
cognitive_data['subject_visit'] = cognitive_data['PATNO'].astype(str) + '_' + cognitive_data['Visit ID'].astype(str)
cognitive_data.drop(columns=['PATNO','Visit ID'], inplace=True)
cognitive_data.head(3)
```

## Data Selection

```python
cognitive_data['UPDRS_III_ProgressionType'] = cognitive_data['UPDRS_III_ProgressionType'].map({'Substantial improv.': -1, 'No significant change.': 0,'Substantial progress.': 1})
cognitive_data['Has Depression']= cognitive_data['Has Depression'].map({True: 1, False: 0})
neuro_cols = ['Motor Exam assessment', 'Coordination assessment', 'Sensory Exam assessment', 'Reflexes assessment', 'CN II-XII assessment']
for col in neuro_cols:
    cognitive_data[col] = cognitive_data[col].map({'Normal': 1, 'Abnormal': 0})

cognitive_data.drop(columns=['UPDRS_I_Class', 'UPDRS_II_Class', 'UPDRS_III_Class',
                             'UPDRS_IV_Class', 'UPDRS_I_ProgressionType', 
                             'UPDRS_II_ProgressionType', 'Perfil_LPA',
                             'MDS-UPDRS Part III Total Score', 'MDS-UPDRS Part IV Total Score'], inplace=True)

numeric_cols = cognitive_data.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = cognitive_data.select_dtypes(include=['object','bool']).columns.tolist()
print("Numeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)
```

## Data Download

```python
cognitive_data.to_csv(ROOT / 'DATA_MODEL_Dev/Cognitive_data_models.csv', index=False)
```
