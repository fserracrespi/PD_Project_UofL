```python
problematic_patnos=updrs2_data.loc[updrs2_data['MDS-UPDRS Part II Total Score'].isna(),'PATNO'].to_list()

# --- Función para rellenar valores faltantes ---
def fill_missing_values(lista_val):
    lista_val = lista_val.copy()
    for i in range(1, len(lista_val)):
        if np.isnan(lista_val[i]):
            lista_val[i] = lista_val[i-1]
    for i in range(len(lista_val)-2, -1, -1):
        if np.isnan(lista_val[i]):
            lista_val[i] = lista_val[i+1]
    return lista_val


# --- Aplicar la función a cada PATNO ---

for patno in problematic_patnos:
    mask = updrs2_data['PATNO'] == patno
    valores_originales = updrs2_data.loc[mask, 'MDS-UPDRS Part II Total Score'].tolist()
    valores_rellenos = fill_missing_values([float(v) if pd.notna(v) else np.nan for v in valores_originales])
    updrs2_data.loc[mask, 'MDS-UPDRS Part II Total Score'] = valores_rellenos

updrs2_data.isna().sum()
```

# FINAL UPDRS 

```python
UPDRS=updrs1Q_data.merge(updrs2_data, on=['PATNO', 'Visit ID'], how='inner')
UPDRS=UPDRS.merge(updrs3_data, on=['PATNO', 'Visit ID'], how='inner')
print(UPDRS.shape)
UPDRS=UPDRS.merge(updrs4_data, on=['PATNO', 'Visit ID'], how='left')
print(UPDRS.shape)
UPDRS.isna().sum()
UPDRS.dropna(how='any', inplace=True)
UPDRS.isna().sum()

```

```python
UPDRS['DBS_Transition_Visit'] = False

for id in UPDRS['PATNO'].unique():
    data_patient = UPDRS.loc[UPDRS['PATNO'] == id].sort_values('Visit ID')
    vals = set(data_patient['Does participant have DBS'].astype(str))
    visits = data_patient['Visit ID'].values
    dbs_vals = data_patient['Does participant have DBS'].astype(int).values

    # Caso 1: todos 0 → nunca tuvo DBS
    if all(x == '0' for x in data_patient['Does participant have DBS'].tolist()):
        UPDRS.loc[UPDRS['PATNO'] == id, 'DBS_Transition_Visit'] = False

    # Caso 2: todos 1 → siempre tuvo DBS
    if all(x == '1' for x in data_patient['Does participant have DBS'].tolist()):
        # Marcar la primera visita como transición
        first_visit = visits[0]
        UPDRS.loc[(UPDRS['PATNO'] == id) & (UPDRS['Visit ID'] == first_visit), 'DBS_Transition_Visit'] = True
        print(f'Patient {id} has always had DBS, transition at visit {first_visit}, length {len(dbs_vals)}.')

    # Caso 3: tiene transición 0 → 1
    if {'0', '1'}.issubset(vals):
        trans_idx = next((i for i in range(1, len(dbs_vals)) if dbs_vals[i-1] == 0 and dbs_vals[i] == 1), None)
        if trans_idx is not None:
            transition_visit = visits[trans_idx]
            UPDRS.loc[(UPDRS['PATNO'] == id) & (UPDRS['Visit ID'] == transition_visit), 'DBS_Transition_Visit'] = True

        print(f'Patient {id} has mixed DBS status (transition at visit {transition_visit}), length {len(dbs_vals)}.')
```

```python
for id in UPDRS['PATNO'].unique():
    data_patient = UPDRS.loc[UPDRS['PATNO'] == id].sort_values('Visit ID').copy()
    trans_flags = data_patient['DBS_Transition_Visit'].values
    visits = data_patient['Visit ID'].values

    # Si no hay transición, todo 0
    if not any(trans_flags):
        UPDRS.loc[UPDRS['PATNO'] == id, 'DBS_Post_Transition'] = 0
        continue

    # Si hay transición → buscar índice
    trans_idx = np.where(trans_flags)[0][0]

    # Asignar número de visitas después (entero)
    for j in range(len(visits)):
        post_val = max(0, j - trans_idx)
        UPDRS.loc[
            (UPDRS['PATNO'] == id) & (UPDRS['Visit ID'] == visits[j]),
            'DBS_Post_Transition'
        ] = int(post_val)
```
