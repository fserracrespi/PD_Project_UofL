### Combinations of modalities in V04

```python

# =========================
# Diccionario de modalidades
# =========================
dict_modal_subjects = {
    'DaTScan': datscan_v04['Subject ID'].unique(),
    'fMRI': fmri_v04['Subject ID'].unique(),
    'DTI': dti_v04['Subject ID'].unique(),
    'MRI_PD': mri_PD_v04['Subject ID'].unique(),
    'MRI_T1': mri_T1_v04['Subject ID'].unique(),
    'MRI_T2': mri_T2_v04['Subject ID'].unique(),
    'PET': pet_v04['Subject ID'].unique(),
    'CT': ct_v04['Subject ID'].unique()
}

dict_images_subjects = {
    'DaTScan': datscan_v04['Subject ID'].shape[0],
    'fMRI': fmri_v04['Subject ID'].shape[0],
    'DTI': dti_v04['Subject ID'].shape[0], 
    'MRI_PD': mri_PD_v04['Subject ID'].shape[0],
    'MRI_T1': mri_T1_v04['Subject ID'].shape[0],
    'MRI_T2': mri_T2_v04['Subject ID'].shape[0],  # modalidad vacía
    'PET': pet_v04['Subject ID'].shape[0],
    'CT': ct_v04['Subject ID'].shape[0]
}

list_modalities = list(dict_modal_subjects.keys())

# =========================
# DataFrame 1: intersecciones entre modalidades (MultiIndex)
# =========================
combinations_keys = list(combinations(list_modalities, 2))
index = pd.MultiIndex.from_tuples(
    combinations_keys,
    names=['Modality_1', 'Modality_2']
)
df = pd.DataFrame(index=index, columns=['V04','V06','V08','V10'], dtype=int)

for comb in combinations_keys:
    set_1 = set(dict_modal_subjects[comb[0]])
    set_2 = set(dict_modal_subjects[comb[1]])
    common_subjects = set_1.intersection(set_2)
    print(f'Intersection {comb[0]} & {comb[1]}: {len(common_subjects)} subjects')
    df.loc[(comb[0], comb[1]), 'V04'] = len(common_subjects)

# =========================
# DataFrame 2: total de sujetos por modalidad
# =========================
df2 = pd.DataFrame(index=list_modalities, columns=['V04','V06','V08','V10'], dtype=int)
df_total_images = pd.DataFrame(index=list_modalities, columns=['Total_Images_PD_V04', 'Total_Images_PD_V06', 'Total_Images_PD_V08', 'Total_Images_PD_V10', 'Total_Images_PD_V12'], dtype=int)
for modality in list_modalities:
    set_modality = set(dict_modal_subjects[modality])
    print(f'Total subjects in {modality}: {len(set_modality)}')
    df2.loc[modality, 'V04'] = len(set_modality)
    df_total_images.loc[modality,'Total_Images_PD_V04'] = dict_images_subjects[modality]


```

```python
dict_modal_subjects = {'DaTScan':[],
                        'fMRI':[],
                        'DTI':[],
                        'MRI_PD':[],
                        'MRI_T1':[],
                        'MRI_T2':[],
                        'PET':pet_v12['Subject ID'].unique(),
                        'CT':ct_v12['Subject ID'].unique()}

dict_images_subjects = {'DaTScan':0,
                        'fMRI':0,
                        'DTI':0,
                        'MRI_PD':0,
                        'MRI_T1':0,
                        'MRI_T2':0,
                        'PET':pet_v12['Subject ID'].shape[0],
                        'CT':ct_v12['Subject ID'].shape[0]}

list_modalities = list(dict_modal_subjects.keys())

# =========================
# DataFrame 1: intersecciones entre modalidades (MultiIndex)
# =========================
combinations_keys = list(combinations(list_modalities, 2))
for comb in combinations_keys:
    set_1 = set(dict_modal_subjects[comb[0]])
    set_2 = set(dict_modal_subjects[comb[1]])
    common_subjects = set_1.intersection(set_2)
    print(f'Intersection {comb[0]} & {comb[1]}: {len(common_subjects)} subjects')
    df.loc[(comb[0], comb[1]), 'V12'] = len(common_subjects)

# =========================
# DataFrame 2: total de sujetos por modalidad
# =========================
for modality in list_modalities:
    set_modality = set(dict_modal_subjects[modality])
    print(f'Total subjects in {modality}: {len(set_modality)}')
    df2.loc[modality, 'V12'] = len(set_modality)
    df_total_images.loc[modality,'Total_Images_PD_V12'] = dict_images_subjects[modality]

```
